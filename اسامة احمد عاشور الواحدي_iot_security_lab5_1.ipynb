```python
import time
import json
import random
from datetime import datetime

def now_ts():
    return datetime.utcnow().strftime("%Y-%m-%d %H:%M:%S UTC")

def log(stage, msg):
    print(f"[{now_ts()}] [{stage}] {msg}")

def simulate_sensor_readings(n=5, seed=123):
    random.seed(seed)
    readings = []
    for i in range(n):
        temp = round(random.uniform(18.0, 35.0), 2)
        hum  = round(random.uniform(20.0, 80.0), 2)
        readings.append({"timestamp": now_ts(), "temperature": temp, "humidity": hum})
        time.sleep(0.02)
    return readings

def xor_encrypt(data_bytes, key):
    encrypted = bytes([b ^ key for b in data_bytes])
    return encrypted

def xor_decrypt(enc_bytes, key):
    decrypted = bytes([b ^ key for b in enc_bytes])
    return decrypted

log("Part I", "Starting sensor simulation and encryption.")
sensor_data = simulate_sensor_readings(n=4, seed=2025)
log("Part I", f"Generated readings:\n{json.dumps(sensor_data, indent=2)}")

key = random.randint(1, 255)
log("Part I", f"XOR key: {key}")

payloads = []
for r in sensor_data:
    plaintext = json.dumps(r).encode("utf-8")
    ct = xor_encrypt(plaintext, key)
    payloads.append(ct)
    log("Part I", f"Encrypted size: {len(ct)} bytes")

decrypted = []
for ct in payloads:
    pt = xor_decrypt(ct, key)
    obj = json.loads(pt.decode("utf-8"))
    decrypted.append(obj)

log("Part I", f"Decrypted readings:\n{json.dumps(decrypted, indent=2)}")
```
