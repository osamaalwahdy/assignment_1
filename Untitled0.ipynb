```python
import math

print("=== RSA Key Generation ===")

p = 7
q = 11
print("p =", p)
print("q =", q)

n = p * q
print("n = p * q =", n)

phi = (p - 1) * (q - 1)
print("φ(n) =", phi)

e = 17
if math.gcd(e, phi) != 1:
    raise ValueError("e and φ(n) must be coprime!")
print("Selected Public Key exponent e =", e)

d = pow(e, -1, phi)
print("Private Key d =", d)

print("\nPublic Key  = (e =", e, ", n =", n, ")")
print("Private Key = (d =", d, ", n =", n, ")")

def encrypt(m, e, n):
    return pow(m, e, n)

def decrypt(c, d, n):
    return pow(c, d, n)

print("\n=== Encryption Phase ===")
m1 = 2
m2 = 3
print("m1 =", m1)
print("m2 =", m2)

c1 = encrypt(m1, e, n)
c2 = encrypt(m2, e, n)
print("E(m1) =", c1)
print("E(m2) =", c2)

print("\n=== Homomorphic Property Test ===")
cipher_product = (c1 * c2) % n
print("E(m1) * E(m2) mod n =", cipher_product)

decrypted_result = decrypt(cipher_product, d, n)
print("Decrypt( E(m1)*E(m2) mod n ) =", decrypted_result)

expected = (m1 * m2) % n
print("Expected (m1 * m2 mod n) =", expected)

if decrypted_result == expected:
    print("Homomorphic Property Verified ✓")
else:
    print("Homomorphic Property FAILED ✗")

print("\n=== Final Log (for PDF Report) ===")
print(f"p = {p}, q = {q}")
print(f"n = {n}, φ(n) = {phi}")
print(f"Public key (e, n) = ({e}, {n})")
print(f"Private key (d, n) = ({d}, {n})")
print(f"m1 = {m1}, m2 = {m2}")
print(f"E(m1) = {c1}, E(m2) = {c2}")
print(f"E(m1)*E(m2) mod n = {cipher_product}")
print(f"Decrypted result = {decrypted_result}")
print(f"Expected = {expected}")
```
